# Project 2: Statistical Plots & Distribution Analysis

## Objective
Analyze data distributions using histograms, KDEs, and boxplots, compare groups, detect outliers, and interpret skewness.

## Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Dataset

```python
df = sns.load_dataset('titanic')
df.head()
```

## Histogram: Age Distribution

```python
plt.savefig("age_distribution.png", dpi=300, bbox_inches="tight")
```

```python
plt.figure()
plt.hist(df['age'].dropna(), bins=30)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

## KDE Plot: Fare Distribution

```python
plt.figure()
sns.kdeplot(df['fare'].dropna())
plt.title('Fare Distribution (KDE)')
plt.xlabel('Fare')
plt.show()
```

## Boxplot: Fare Outliers

```python
plt.savefig("age_distribution.png", dpi=300, bbox_inches="tight")

```

```python
plt.figure()
sns.boxplot(x=df['fare'])
plt.title('Fare Boxplot')
plt.show()
```

## Group-wise Comparison: Age by Gender

```python
plt.figure()
sns.boxplot(x='sex', y='age', data=df)
plt.title('Age Distribution by Gender')
plt.show()
```

## Group-wise Comparison: Age by Class

```python
plt.figure()
sns.kdeplot(data=df, x='age', hue='class')
plt.title('Age Distribution by Passenger Class')
plt.show()
```

## Skewness Analysis

```python
df[['age', 'fare']].skew()
```

```python
df.to_csv("titanic_cleaned.csv", index=False)
```

```python
skew = df[['age', 'fare']].skew()

with open("skewness_results.txt", "w") as f:
    f.write(str(skew))
```

## Final Interpretation
Passenger age is moderately right-skewed, with most travelers between 20 and 40 years. Fare distribution is highly right-skewed, indicating a small number of very high-paying passengers. First-class passengers tend to be older and pay higher fares, while third-class passengers are younger with lower fares. Boxplots highlight clear fare outliers, and KDE plots reveal smooth distribution patterns.
